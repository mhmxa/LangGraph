```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
```

```python
load_dotenv()

model = ChatGoogleGenerativeAI(model="gemini-3.6-flash")

class LLMState(TypedDict):

    question: str
    answer: str
```

```python
def llm_qa(state: LLMState) -> LLMState:

    question = state["question"]
    prompt = f"Answer the following question. \n{question}"
    answer = model.invoke(prompt).content
    state["answer"] = answer[0]['text']

    return state
```

```python
graph = StateGraph(LLMState)

graph.add_node("LLM QA", llm_qa)

graph.add_edge(START, "LLM QA")
graph.add_edge("LLM QA", END)

workflow = graph.compile()
```

```python
initial_state = {'question': "Who has the best armpits of female character in rezero anime"}
final_state = workflow.invoke(initial_state)
print(final_state)
```
